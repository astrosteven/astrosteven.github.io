## 1. Temperature of the Sun.

First, you're going to calculate the effective temperature of the Sun. As you hopefully recall, the Sun and other stars behave like black-bodies. This means that their radiation can be described in terms of a characteristic temperature. Then, you can relate the Sun's luminosity, $L$, to is radius, $R$, and effective temperature, $T_{eff}$ using the Stefan-Boltzmann law: 

(1) $L = 4 \pi R^2 \sigma T_{eff}^4$,

where $\sigma$ is a constant $\sigma = 5.670367 \dot 10^{−8}$ W m$^{-2}$K$^{-4}$. (Check: verify that these units make sense.)
