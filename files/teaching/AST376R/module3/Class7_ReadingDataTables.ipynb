Above you can see there are several useful options you can use with `np.loadtxt`.<br>
A few of the most common ones are: <br>
> `unpack=True` $-$ this will trasnpose the columns into numpy arrays. <br>
> `dtype=[options]` $-$ set the data-type of the resulting array; default is float. Use str if reading in strings. <br>
> `usecols=(0,)` $-$ reads only the first column. Can list more columns to read as many columns as desired. 

Below are some examples:

````python
# Loads the file named 'File1.dat' 
# Transposes the columns of the file into arrays and reads them as float
xf, yf = np.loadtxt('File1.dat',unpack=True, dtype=float) 

# Transposes the columns of the file into arrays and reads them as strings
xs, ys = np.loadtxt('File1.dat',unpack=True, dtype=str) 

# Reads only the first column with usecols=(0,), transposes it into an array and reads it as float
x1 = np.loadtxt('File1.dat',unpack=True,dtype=float,usecols=(0,))
````

````python
#Lets examine these arrays
print('xf =', xf)
print('yf =', yf)
print('xs =', xs)
print('ys =', ys)
print('x1 =', x1)
````

# Writing Files:

Writing the contents of arrays to a file is another common use for Python. <br> 
An example is shown below: <br>
Two arrays, nx and ny, are created, transposed them from rows to columns with n`p.c_`, and saved to a file using `np.savetxt()`:<br>

```python
#  np.arange(a,b,c) create an array starting at a, ending at b (not included), with steps of c:
nx = np.arange(1, 15, 0.5)
ny = np.arange(21, 35, 0.5)

# np.c_: transposes 2 arrays into 2 columns:
table = np.c_[nx, ny]

# Save the data table as a file named 'output1.txt':
np.savetxt('Output.txt', np.c_[nx, ny], fmt='%6.2f')
np.savetxt('Output_int.txt', np.c_[nx, ny], fmt='%6.4i')
np.savetxt('Output_str.txt', np.c_[nx, ny], fmt='%s')
```

`fmt` gives the format for the outputted values. <br>
The syntax is to give the *data type* specifier preceded by the `\%` character. <br>
`fmt='%6.2f'` means each column in the new text file is defined by: <br>
- f: is a float <br>
- 6: \# of fixed width spaces (specified by first number, counting any decimal point) <br>
- 2: \# of digits past the decimal point (second number) <br> 

`fmt='%6.4i'` means each column in the new text file is defined by: <br>
- i: is an integer <br>
- 6: \# of fixed width spaces (specified by first number) <br>
- 4: \# of integer digits (second number) <br>  

`fmt='%s'` means each column in the new text file is defined by: <br>
- s: is a string <br>

## Writing Files with Pandas

Saving a `DataFrame` in `Pandas` is incredibly easy. <br>
You can save `Pandas DataFrames` to a number of different formatted outputs using the following commands: <br>


| File Type | Pandas Command                | Notes                                |
|:-         |:-                             |:-                                    |
| `.csv`    | `pd.DF.to_csv('file_name')`   | For comma separated formated output  |
| `.txt`    | `pd.DF.to_table('file_name')` | For tab delimited formated output    |
| `.xls`    | `pd.DF.to_excel('file_name')` | For excel formated output            |
| `html`    | `pd.DF.to_html('file_name')`  | For html formated output             |
| `sql`     | `pd.DF.to_sql('file_name')`   | For structured query language output |
| `.tex`    | `pd.DF.to_latex('file_name')` | For latex formated output            |


Example: <br>
To save a subset of an existing DataFrame to a new file, say the last 20 rows of the data frame, use: <br>
```python
# First, read in the full data frame from the input file
df = pd.read_csv('Finkelstein24_data.csv')

# Define last 20 rows of DataFrame using indexing
df_last20 = df[-20:]

# Write this new DataFrame into a csv file called "last20rows.csv"
df_last20.to_csv('_last20rows.csv',index_label='Index')
```

## Extracting data from Dataframes 

### np.where()
`np.where()` can be used with `Pandas Dataframe` in a similar way as with arrays. <br>
It will extract the indices of the elements of the column that meet the given condition. <br>
The indices can then be used with the `df.loc[]` function to extract the part of the `dataframe` that meets the condition. <br>
The conditions can be applied directly to the `df` as well to filter the data.  <br>

````python
# Read in the .csv file:
df = pd.read_csv('Finkelstein24_data.csv') 

# Extract the column labeled 'photz':
photz = df['photz']

# Extracts the indices of photz where photz is greater than 11.0:
condition = photz > 11.0
indices   = np.where(condition)

# Prints the indices:
print("Indices of where photz > 11.0 = ", indices[0])
 
# Extract the values from the .csv file for the indices meeting the condition above:
subdata = df.loc[indices]

# Print the extracted data from the csv file:
print("Data where photz > 11.0 = ")
print(subdata)
````


### Python masking

We can also extract data from `Pandas Dataframes` using standard `Python` masks instead of `np.where`.  <br>
`Python` masks are defined using hard brackets [].  <br>
Masking is another option for extracting subsets of data, and this kind of masking can be used with `DataFrames` and `numpy arrays`, but not with `lists`.

For example, 
```python
over11 = df[df['photz']>11]
```
will extract all rows of data from the dataframe where the photz is larger than 11.0

Each column extracted from `df` can be treated like an array, so usual array functions like 
```python
df['photz'].max() 
df['photz'].min()
df['photz'].mean()
```
can be used on them if the column data type are int or float. <br>
Mathematical functions can be applied to them as well. <br>
